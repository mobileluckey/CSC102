````python
# Cognitive AI Integration - Google Colab Starter
# Criterion 1: Create and integrate cognitive artificial intelligence into applications

"""
This notebook demonstrates integrating cognitive AI (LLMs) into applications.
We'll create a simple chatbot application with memory and context awareness.
"""

# Install required libraries
!pip install anthropic openai -q

import anthropic
import json
from datetime import datetime

# =============================================================================
# SETUP: Add your API key here
# Get free API key from: https://console.anthropic.com/
# =============================================================================
ANTHROPIC_API_KEY = "your-api-key-here"  # Replace with your actual key

# =============================================================================
# Example 1: Basic Cognitive AI Integration
# =============================================================================

def simple_ai_chat(user_message, api_key):
    """
    Basic integration of Claude AI into an application
    """
    client = anthropic.Anthropic(api_key=api_key)

    response = client.messages.create(
        model="claude-sonnet-4-20250514",
        max_tokens=1024,
        messages=[
            {"role": "user", "content": user_message}
        ]
    )

    return response.content[0].text

# =============================================================================
# Example 2: Cognitive AI with Memory (Conversation History)
# =============================================================================

class CognitiveAssistant:
    """
    A cognitive AI assistant with conversation memory and context awareness
    """
    def __init__(self, api_key, system_prompt="You are a helpful AI assistant."):
        self.client = anthropic.Anthropic(api_key=api_key)
        self.system_prompt = system_prompt
        self.conversation_history = []

    def chat(self, user_message):
        """Send message and maintain conversation context"""
        # Add user message to history
        self.conversation_history.append({
            "role": "user",
            "content": user_message
        })

        # Get AI response with full context
        response = self.client.messages.create(
            model="claude-sonnet-4-20250514",
            max_tokens=1024,
            system=self.system_prompt,
            messages=self.conversation_history
        )

        # Extract response text
        assistant_message = response.content[0].text

        # Add assistant response to history
        self.conversation_history.append({
            "role": "assistant",
            "content": assistant_message
        })

        return assistant_message

    def clear_history(self):
        """Reset conversation memory"""
        self.conversation_history = []

    def get_conversation_summary(self):
        """Get summary of conversation"""
        return f"Messages exchanged: {len(self.conversation_history)}"

# =============================================================================
# Example 3: Task-Specific Cognitive AI Integration
# =============================================================================

class CodeReviewAssistant:
    """
    Specialized cognitive AI for code review tasks
    """
    def __init__(self, api_key):
        self.client = anthropic.Anthropic(api_key=api_key)

    def review_code(self, code, language="python"):
        """Analyze code and provide cognitive insights"""
        prompt = f"""Review this {language} code and provide:
1. Code quality assessment
2. Potential bugs or issues
3. Performance optimization suggestions
4. Best practice recommendations

Code:
```{language}
{code}
```
"""

        response = self.client.messages.create(
            model="claude-sonnet-4-20250514",
            max_tokens=2000,
            messages=[{"role": "user", "content": prompt}]
        )

        return response.content[0].text

# =============================================================================
# Example 4: Multi-Step Cognitive Reasoning
# =============================================================================

class ProblemSolver:
    """
    Cognitive AI that breaks down complex problems into steps
    """
    def __init__(self, api_key):
        self.client = anthropic.Anthropic(api_key=api_key)

    def solve_problem(self, problem_description):
        """Use cognitive reasoning to solve complex problems"""
        prompt = f"""Analyze this problem using step-by-step cognitive reasoning:

Problem: {problem_description}

Please:
1. Break down the problem into components
2. Analyze each component
3. Propose solutions
4. Evaluate trade-offs
5. Recommend best approach
"""

        response = self.client.messages.create(
            model="claude-sonnet-4-20250514",
            max_tokens=2000,
            messages=[{"role": "user", "content": prompt}]
        )

        return response.content[0].text

# =============================================================================
# TESTING THE EXAMPLES
# =============================================================================

if __name__ == "__main__":
    print("=" * 70)
    print("COGNITIVE AI INTEGRATION DEMONSTRATIONS")
    print("=" * 70)

    # Note: Replace ANTHROPIC_API_KEY with your actual key to run these examples

    # Test 1: Simple Integration
    print("\n[1] SIMPLE AI CHAT")
    print("-" * 70)
    print("Example: simple_ai_chat('What is machine learning?', ANTHROPIC_API_KEY)")

    # Test 2: Conversational AI with Memory
    print("\n[2] COGNITIVE ASSISTANT WITH MEMORY")
    print("-" * 70)
    print("""Example:
assistant = CognitiveAssistant(ANTHROPIC_API_KEY)
response1 = assistant.chat("My name is John")
response2 = assistant.chat("What's my name?")  # AI remembers context
""")

    # Test 3: Specialized Task AI
    print("\n[3] CODE REVIEW ASSISTANT")
    print("-" * 70)
    print("""Example:
reviewer = CodeReviewAssistant(ANTHROPIC_API_KEY)
review = reviewer.review_code('''
def calculate(x, y):
    return x + y / 2
''')
""")

    # Test 4: Multi-Step Reasoning
    print("\n[4] PROBLEM SOLVER")
    print("-" * 70)
    print("""Example:
solver = ProblemSolver(ANTHROPIC_API_KEY)
solution = solver.solve_problem(
    "Design a recommendation system for an e-commerce platform"
)
""")

    print("\n" + "=" * 70)
    print("To run these examples:")
    print("1. Get API key from https://console.anthropic.com/")
    print("2. Replace 'your-api-key-here' with your actual key")
    print("3. Run the code blocks above")
    print("=" * 70)

# =============================================================================
# INTEGRATION INTO APPLICATIONS
# =============================================================================

"""
REAL-WORLD APPLICATION EXAMPLES:

1. Customer Support Chatbot:
   - Use CognitiveAssistant with custom system prompt
   - Integrate with web interface (Streamlit/Flask)
   - Add database for conversation persistence

2. Content Analysis Tool:
   - Process documents with cognitive understanding
   - Extract insights, summaries, key points
   - Classify and categorize information

3. Educational Tutor:
   - Personalized learning assistance
   - Adaptive responses based on student level
   - Track progress over conversation history

4. Development Assistant:
   - Code generation and review
   - Documentation generation
   - Bug detection and fixing suggestions

KEY CONCEPTS DEMONSTRATED:
✓ API integration with cognitive AI
✓ Context management and memory
✓ Task-specific AI applications
✓ Multi-step reasoning and problem solving
✓ Error handling and response processing
"""
````
